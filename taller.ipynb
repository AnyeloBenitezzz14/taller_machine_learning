```python
import pandas as pd
import numpy as np
import unicodedata
from sklearn.preprocessing import MinMaxScaler

# Cargar el archivo original ubicado en la misma carpeta del notebook
df = pd.read_csv("Data/datos_para_limpiar - Original.csv")

print("Tamaño inicial (filas, columnas):", df.shape)
df.head()
```

Luego se eliminan datos duplicados o innecesarios como id y notas

```python
df = df.drop(columns=["ID", "Notas"], errors="ignore")
print("Columnas actuales:", df.columns.tolist())
```

Luego normalizamos los datos, quitando espacios, tildes y cosas del estilo

```python
def quitar_tildes(texto):
    texto = unicodedata.normalize("NFKD", str(texto))
    return "".join(c for c in texto if not unicodedata.combining(c))

df.columns = [
    quitar_tildes(col).strip().lower().replace(" ", "_")
    for col in df.columns
]
print("Nombres normalizados:", df.columns.tolist())
```

```python
def quitar_tildes(texto):
    texto = unicodedata.normalize("NFKD", str(texto))
    return "".join(c for c in texto if not unicodedata.combining(c))

df.columns = [
    quitar_tildes(col).strip().lower().replace(" ", "_")
    for col in df.columns
]
print("Nombres normalizados:", df.columns.tolist())
```

luego se normalizan las ciudades usando el tipo de dato anterior de quitar tildes, para poder ver tipos de dato que no aparecen

```python
def normalizar_ciudad(valor):
    if pd.isna(valor):
        return np.nan
    texto = quitar_tildes(str(valor)).strip().lower()
    texto = " ".join(texto.split())
    equivalencias = {
        "bogota": "Bogotá",
        "medellin": "Medellín",
        "cali": "Cali"
    }
    return equivalencias.get(texto, str(valor).strip())

if "ciudad" in df.columns:
    df["ciudad"] = df["ciudad"].apply(normalizar_ciudad)
    print(df["ciudad"].value_counts(dropna=False))
```

 Preparamos los tipos y tratamos los valores nulos
Primero se convierten las columnas numéricas necesarias para calcular correctamente la media y la mediana. Para edad, la media se calcula únicamente con valores entre 0 y 120

```python
# Convertir columnas numéricas; los valores que no se puedan convertir pasan a NaN
for col in ["edad", "salario", "ingresos", "gastos"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Edad: media calculada solo con edades válidas (0 a 120)
if "edad" in df.columns:
    edades_validas = df.loc[df["edad"].between(0, 120), "edad"]
    media_edad = edades_validas.mean()
    if pd.notna(media_edad):
        df["edad"] = df["edad"].fillna(media_edad)
    print("Media de edad válida usada:", media_edad)

# Salario: reemplazar nulos por la mediana
if "salario" in df.columns:
    mediana_salario = df["salario"].median()
    if pd.notna(mediana_salario):
        df["salario"] = df["salario"].fillna(mediana_salario)
    print("Mediana de salario usada:", mediana_salario)

# Ciudad y género: reemplazar nulos por la moda
for col in ["ciudad", "genero"]:
    if col in df.columns:
        modas = df[col].mode(dropna=True)
        if not modas.empty:
            df[col] = df[col].fillna(modas.iloc[0])
            print(f"Moda usada para {col}:", modas.iloc[0])

print("\nNulos después del tratamiento:")
display(df.isna().sum().to_frame("valores_nulos"))
```

```python
if "edad" in df.columns:
    df["edad"] = pd.to_numeric(df["edad"], errors="coerce")

if "fecha" in df.columns:
    df["fecha"] = pd.to_datetime(df["fecha"], errors="coerce", dayfirst=True)

df.info()
if "fecha" in df.columns:
    display(df[["fecha"]].head())
```

Convertimos los datos a su tipo de dato correspondiente y se aplican filtros para los datos invalidos 

```python
filas_antes = len(df)

if "edad" in df.columns:
    df = df[df["edad"].isna() | df["edad"].between(0, 120)]

if "salario" in df.columns:
    df = df[df["salario"].isna() | (df["salario"] <= 1_000_000_000)]

print("Registros eliminados por filtros:", filas_antes - len(df))
print("Tamaño actual:", df.shape)
```

luego definimos los valores para medir el nivel 

```python
if "nivel" in df.columns:
    mapa_nivel = {"bajo": 1, "medio": 2, "alto": 3}
    nivel_texto = df["nivel"].astype("string").str.strip().str.lower()
    nivel_texto = nivel_texto.replace({"": pd.NA, "n/a": pd.NA, "na": pd.NA})
    df["nivel"] = nivel_texto.map(mapa_nivel)
    print(df["nivel"].value_counts(dropna=False))
```

transformamos los datos genero y ciudad en valores numericos y eleminamos una variable de referecnia

```python
categoricas = [col for col in ["genero", "ciudad"] if col in df.columns]
if categoricas:
    df = pd.get_dummies(df, columns=categoricas, drop_first=True, dtype=int)

display(df.head())
print("Columnas después de codificar:", df.columns.tolist())
```

Aca definimos la utilidad que es ingresos-gastos, ya luego escalamos eda e ingresos


```python
if {"ingresos", "gastos"}.issubset(df.columns):
    df["utilidad"] = df["ingresos"] - df["gastos"]
else:
    print("Aviso: no se pudo calcular utilidad porque falta ingresos o gastos.")

columnas_escalar = [col for col in ["edad", "ingresos"] if col in df.columns]
if columnas_escalar:
    scaler = MinMaxScaler()
    # MinMaxScaler requiere datos no nulos; transforma solo filas con valores disponibles
    for col in columnas_escalar:
        validos = df[col].notna()
        if validos.any():
            df.loc[validos, col] = scaler.fit_transform(df.loc[validos, [col]]).ravel()

display(df[ [c for c in ["edad", "ingresos", "gastos", "utilidad"] if c in df.columns] ].head())
```

usando filas_antes eliminamos duplicados


```python
filas_antes = len(df)
df = df.drop_duplicates().copy()
print("Duplicados eliminados:", filas_antes - len(df))
print("Tamaño final antes de exportar:", df.shape)
```

por ultimo realizamos un resumen y guardamos el dataset limpio


```python
print("Resumen final:")
df.info()

print("\nNulos finales por columna:")
display(df.isna().sum().to_frame("valores_nulos"))

display(df.head())

df.to_csv("datos_limpios.csv", index=False, encoding="utf-8-sig")
print("Archivo guardado como datos_limpios.csv")
```
